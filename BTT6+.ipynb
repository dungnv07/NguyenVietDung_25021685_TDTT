```python
a = map(int, input().split())
print(list(sorted(set(a))))
```

```python
a = map(int, input().split())
s = 0
l = []
for i in a:
    s += i
    l.append(s)
print(l)
```

```python
a = tuple(map(int, input().split()))
k = int(input())
m = list(a)
for i in range(k):
    x = m.pop(0)
    m.append(x)
print(tuple(m))
```

```python
a = input().split(' ')
d = {}
for p in a:
    k, v = p.split(':')
    if k not in d:
        d[k] = [v]
    else:
        d[k] += [v]
print(d)
```
