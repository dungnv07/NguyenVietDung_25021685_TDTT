```python
def first(a, k):
    if k in a:
        return a.index(k) + 1
    else:
        return -1
a = list(map(int, input().split()))
k = int(input())
print(first(a, k))
```

```python
#W6A24
def push(a):
    for i, v in enumerate(a):
        if v == 0:
            a.pop(i)
            a.insert(0, v)
    return a
a = list(map(int, input().split()))
print(push(a))
```
